# Q1. 7로 나누어떨어지고 5의 배수가 아닌 숫자 찾기

**난이도:** 초급 (Beginner)

100부터 300 사이(양 끝 값 포함)의 숫자 중 **7로 나누어떨어지지만 5의 배수는 아닌** 모든 숫자를 찾는 파이썬 프로그램을 작성하세요. 조건에 맞는 숫자들은 쉼표(`,`)로 구분하여 한 줄로 출력되어야 합니다.

---

### 입력 예시
```python
find_numbers(100, 200)
```

### 출력 예시
```Plaintext
112,119,126,133,147,154,161,168,182,189,196
```

#### 힌트
지정된 범위를 반복할 때 range(시작값, 끝값) 함수를 사용하세요.

````python
# 100부터 300까지 숫자 확인
for n in range(100, 301):
    # 7로 나누어떨어지고 5로 나누었을 때 나머지가 0이 아니면 출력
    if n % 7 == 0 and n % 5 != 0:
        print(n, end=",")#end =","숫자 나열을 정리하는 !
````


# Q2. 정수 제곱 딕셔너리 생성하기

**난이도:** 초급 (Beginner)

정수 $n$을 입력받아 1부터 $n$까지(양 끝 값 포함)의 모든 정수 $i$에 대해 $(i, i^2)$ 쌍을 포함하는 딕셔너리를 생성하는 파이썬 함수를 작성하세요. 작성한 함수는 완성된 딕셔너리를 반환해야 합니다.

---

### 입력 예시
```python
generate_square_dict(8)
```

### 출력 예시
```Python
{1: 1, 2: 4, 3: 9, 4: 16, 5: 25, 6: 36, 7: 49, 8: 64}
```

#### 힌트
dict() 함수를 사용하여 빈 딕셔너리를 만드세요.
반복문을 사용하여 1부터 $n$까지의 범위를 순회하세요.

````python
def generate_square_dict(n):
    result = {}  # 빈 딕셔너리 상자

    for i in range(1, n + 1):  # 1부터 n까지 숫자를 하나씩 꺼내
        result[i] = i * i      # 숫자에 자기 자신을 곱해서 상자에 넣기

    return result

# 8을 넣었을 때
print(generate_square_dict(8))
````

# Q3. D 값을 이용한 Q 값 계산하기

**난이도:** 초급 (Level 1)

다음 공식을 사용하여 $Q$ 값을 계산하는 파이썬 함수를 작성하세요.

$$Q = \sqrt{\frac{2 \cdot C \cdot D}{H}}$$

여기서 $C$는 50, $H$는 30으로 고정됩니다. 함수는 쉼표로 구분된 일련의 $D$ 값 문자열만 입력으로 받습니다. 계산 결과는 가장 가까운 정수로 반올림하고, 쉼표로 구분하여 한 줄로 출력되어야 합니다.

---

### 입력 예시
```python
calculate_q_values("100,150,180")  # 100, 150, 180은 D의 입력값들입니다.
```

### 출력 예시
```Plaintext
18,22,24
```

#### 힌트
제곱근을 계산하려면 math.sqrt() 함수를 사용하세요.
계산을 위해 입력값을 float 형식으로 변환하고, round() 함수를 사용하여 결과를 반올림하세요.

````python
import math  # 제곱근(루트) 계산기 도구

def calculate_q_values(d_string):
    # 쉼표(,)를 기준으로 입력받은 글자들을 하나씩 쪼개요 ("100", "150", "180")
    d_list = d_string.split(",")
    result = []

    for d in d_list:
        d_num = float(d)  # 글자를 계산할 수 있는 숫자로 바꿔요

        # 공식 그대로 계산하기: 루트( (2 * 50 * D) / 30 )
        q = math.sqrt((2 * 50 * d_num) / 30)

        # 반올림한 뒤 문자로 바꿔서 결과 리스트에 넣어준다.
        result.append(str(round(q)))

    # 결과들을 쉼표","로 연결해서 출력
    print(",".join(result))

# 실행
calculate_q_values("100,150,180")
````

# Q4. 행렬 배열 생성하기

**난이도:** 중급 (Level 2)

두 정수 $M$과 $N$을 입력받아 2차원 배열(행렬)을 생성하는 파이썬 프로그램을 작성하세요. 배열의 $i$번째 행과 $j$번째 열에 위치한 값은 $i \times j$가 되어야 합니다 (인덱스는 0부터 시작).

---

### 입력 예시
```python
create_matrix(4, 3)
```

### 출력 예시
```Python
[[0, 0, 0],
 [0, 1, 2],
 [0, 2, 4],
 [0, 3, 6]]
```

예: 숫자 6은 3번째 행(row=3), 2번째 열(column=2)에 위치합니다. (3 * 2 = 6)

#### 힌트
중첩된 리스트 컴프리헨션(List Comprehension)을 사용하여 2차원 행렬을 생성하세요.

````python
def create_matrix(M, N):
    matrix = [[i * j for j in range(N)] for i in range(M)]
    return matrix

# 1. 행렬을 만들어요
result = create_matrix(4, 3)

# 2. 첫 줄, 중간 줄, 마지막 줄의 띄어쓰기를 딱 맞춰서 출력해요
print("[" + str(result[0])+"," )
for row in result[1:-1]:
    print(" " + str(row)+"," )
print(" " + str(result[-1]) + "]")
````

# Q5. 고유 단어 정렬하기

**난이도:** 중급 (Level 2)

공백으로 구분된 단어들의 시퀀스를 입력받아, 중복을 제거하고 알파벳순으로 정렬된 단어 문자열을 반환하는 파이썬 프로그램을 작성하세요.

---

### 입력 예시
```python
unique_sorted_words('dog cat apple cat banana dog')
```

### 출력 예시
```Python
'apple banana cat dog'
```

#### 힌트
중복을 제거할 때는 set()을 사용하고, 단어를 정렬할 때는 sorted() 함수를 사용하세요.

````python
def unique_sorted_words(text):
    words = text.split()          # 1. 띄어쓰기로 단어들을 쪼개요
    unique = set(words)           # 2. set()으로 중복된 단어를 지워요
    sorted_words = sorted(unique) # 3. 알파벳 순서대로 줄을 세워요
    return " ".join(sorted_words) # 4. 다시 한 칸씩 띄워서 합쳐요

# 실행해보기
print(unique_sorted_words ('dog cat apple cat banana dog'))
````

# Q10. 재귀 함수를 이용한 f(n) 계산하기

**난이도:** 중급 (Level 2)

$n > 0$일 때 $f(n) = f(n-1) + 100$이고, $f(0) = 1$을 만족하는 파이썬 재귀 함수 $f(n)$을 작성하세요.

---

### 입력 예시
```python
print(f(5))
```

### 출력 예시
```Plaintext
501
```

#### 힌트
* $f(n)$의 값을 계산하기 위해 재귀 함수(Recursive Function)를 사용하세요.
* Base Case($n=0$일 때의 종료 조건)를 반드시 명시해야 합니다.

````python
def f(n):
    # 멈추는 기준 (Base Case): n이 0이면 계산을 멈추고 1을 내보내요
    if n == 0:
        return 1

    # 자기 자신을 다시 불러서 100을 더해요
    return f(n - 1) + 100

# 실행해보기
print(f(5))
````
